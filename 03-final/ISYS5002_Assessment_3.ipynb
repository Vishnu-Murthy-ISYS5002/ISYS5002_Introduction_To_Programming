```python
import re
from datetime import datetime, timedelta


# [CONSTANTS]
HEADER_WIDTH = 62
OFFICER_LIST_WIDTH = 23
BOX_WIDTH = 85
LOG_CONTACT_WIDTH = 27
MAX_TEXT_LENGTH = 100
DATE_FORMAT = "%d/%m/%Y"
MIN_YEAR = 2000
MAX_QUANTITY = 100000
CHANNELS = ["Email", "Instagram"]
CONTRIBUTION_TYPES = ["Voucher", "Merch"]
NOT_PROVIDED = "Not provided"
MIN_PHONE_DIGITS = 8
MAX_PHONE_DIGITS = 15
CONTACT_LOG = []
MAIN_MENU_WIDTH = 29
NEW_SPONSOR_WIDTH = 23
SPONSOR_LOG = []
CONFIRMED_STAGE = "Confirmed"
DISPLAY_DATE_FORMAT = "%d %b %Y"
COMPLETE_STATUS = "Complete"
CONTACT_SCORE = 1
FOLLOW_UP_SCORE = 2
CONFIRMED_SCORE = 5
OFFICERS = {"1": "Vishnu Frendy", "2": "Anavi Verde", "3": "Erich Corin", "4": "Izy Ray", "5": "Elysha Miles", "6": "Owen Mendoza",}
LOG_CONTACT_FIELDS = {
    ("sponsor", "Sponsor Name", "Enter sponsor name: ", "text"),
    ("channel", "Communication Channel", "Enter communication channel (Email/Instagram): ", "channel"),
    ("date", "Date", "Enter date (dd/mm/yyyy): ", "date"),
    ("type", "Contribution Type", "Enter contribution type (Voucher/Merch): ", "contribution_type"),
    ("quantity", "Quantity", "Enter quantity: ", "quantity"),
    ("event", "Event Name", "Enter event name: ", "text"),
}
NEW_SPONSOR_FIELDS = [
    ("business_name", "Business name", "Enter sponsor business name: ", "text"),
    ("business_type", "Business type", "Enter sponsor business type: ", "text"),
    ("contact_person", "Contact person", "Enter sponsor contact person: ", "text"),
    ("job_title", "Job title", "Enter sponsor job title: ", "text"),
    ("email", "Email address", "Enter sponsor email address (Enter to skip): ", "email"),
    ("phone", "Phone number", "Enter sponsor phone number (Enter to skip): ", "phone"),
    ("initial_contact_date", "Initial contact date", "Enter sponsor initial contact date (dd/mm/yyyy): ", "date"),
]
DEPARTMENT_REQUEST_COLUMNS = [
    ("department", "Department", 15),
    ("title", "Requests", 35),
    ("due", "Due", 15),
    ("status", "Status", 20),
]
OFFICER_STATS = [
    {"officer": "Vishnu Frendy", "contacts": 5, "follow_ups": 2, "confirmed": 2},
    {"officer": "Anavi Verde", "contacts": 4, "follow_ups": 3, "confirmed": 0},
    {"officer": "Izy Ray", "contacts": 2, "follow_ups": 1, "confirmed": 0},
]
RANKING_COLUMNS = [
    ("rank", "Rank", 15),
    ("officer", "Officer", 15),
    ("contacts", "Contact", 15),
    ("follow_ups", "Follow-Ups", 15),
    ("confirmed", "Confirmed", 15),
    ("score", "Score", 15),
]


# [FUNCTION] PAGE HEADER
def print_header():
    print(("=" * HEADER_WIDTH) + "\n")
    print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")


# [FUNCTION] PUBLIC RELATIONS (PR) OFFICER LIST
def pr_officer_list(width=OFFICER_LIST_WIDTH):
    print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")

    # Display The PR Officer List
    border = "-" * width
    print(border)
    print(f"|{'PR Officer List':^{width - 2}}|")
    print(border)

    for number, officer in OFFICERS.items():
        print(f"| {f'{number}. {officer}':{width - 3}}|")

    print(border)


# [FUNCTION] PRINT BOX LINES
def print_box(lines, width=BOX_WIDTH):
    border = "-" * width
    print(border)

    for line in lines:
        print(f"{'|':2}{line:{width - 3}}{'|':2}")

    print(border)


# [FUNCTION] GET VALID INPUT
def get_valid_input(prompt, rule):
    while True:
        try:
            return VALIDATORS[rule](input(prompt).strip())
        except ValueError as error:
            print(error)


# [FUNCTION] COLLECT FIELDS
def collect_fields(fields):
    record = {}

    for key, label, prompt, rule in fields:
        record[key] = get_valid_input(prompt, rule)

    return record


# [FUNCTION] TEXT INPUT VALIDATION
def validate_text(text):
    if not text:
        raise ValueError("This field cannot be blank.")

    if len(text) > MAX_TEXT_LENGTH:
        raise ValueError(f"Please use {MAX_TEXT_LENGTH} characters or fewer.")

    return text


# [FUNCTION] DATE INPUT VALIDATION
def validate_date(text):
    try:
        date = datetime.strptime(text, DATE_FORMAT).date()
    except ValueError:
        raise ValueError("Invalid date. Please use dd/mm/yyyy.") from None

    if date.year < MIN_YEAR:
        raise ValueError(f"Please enter a date from {MIN_YEAR} onwards.")

    if date > get_today():
        raise ValueError("This date cannot be in the future.")

    return date.strftime(DATE_FORMAT)


# [FUNCTION] QUANTITY INPUT VALIDATION
def validate_quantity(text):
    try:
        quantity = int(text)
    except ValueError:
        raise ValueError("Please enter a whole number.") from None

    if not 1 <= quantity <= MAX_QUANTITY:
        raise ValueError(f"Quantity must be between 1 and {MAX_QUANTITY}.")

    return quantity


# [FUNCTION] OPTION INPUT VALIDATION
def validate_option(text, options):
    for option in options:
        if text.lower() == option.lower():
            return option

    raise ValueError(f"Please enter one of these: {', '.join(options)}.")


# [FUNCTION] CHANNEL INPUT VALIDATION
def validate_channel(text):
    return validate_option(text, CHANNELS)


# [FUNCTION] CONTRIBUTION TYPE INPUT VALIDATION
def validate_contribution_type(text):
    return validate_option(text, CONTRIBUTION_TYPES)


# [FUNCTION] EMAIL INPUT VALIDATION
def validate_email(text):
    if not text:
        return NOT_PROVIDED

    if not re.fullmatch(r"[^@\s]+@[^@\s]+\.[^@\s]+", text):
        raise ValueError("Invalid email. Please use a format like name@example.com.")

    return text


# [FUNCTION] PHONE INPUT VALIDATION
def validate_phone(text):
    if not text:
        return NOT_PROVIDED

    digits = re.sub(r"[\s()+-]", "", text)

    if not digits.isdigit() or not MIN_PHONE_DIGITS <= len(digits) <= MAX_PHONE_DIGITS:
        raise ValueError(f"Phone numbers need {MIN_PHONE_DIGITS} to {MAX_PHONE_DIGITS} digits.")

    return text

# [CONSTANT] VALIDATORS
VALIDATORS = {
    "text": validate_text,
    "date": validate_date,
    "quantity": validate_quantity,
    "channel": validate_channel,
    "contribution_type": validate_contribution_type,
    "email": validate_email,
    "phone": validate_phone,
}


# [FUNCTION] GET TODAY'S LOCAL DATE
def get_today():
    return datetime.today().date()


# [FUNCTION] FORMAT DATE
def format_date(date):
    return date.strftime(DISPLAY_DATE_FORMAT)


# [FUNCTION] FORMAT DAYS
def format_days(days):
    return f"{days} day" if days == 1 else f"{days} days"


# [FUNCTION] DATES SET RELATIVE TO TODAY
def days_from_today(days):
    return datetime.today().date() + timedelta(days=days)


# [CONSTANT] FOLLOW UPS
FOLLOW_UPS = [
    {
        "sponsor": "Kenidonuts",
        "owner": "Vishnu Frendy",
        "channel": "Email",
        "stage": "Contacted",
        "last_contact": days_from_today(-7),
    },
    {
        "sponsor": "Sedap Place",
        "owner": "Vishnu Frendy",
        "channel": "Email",
        "stage": "Confirmed",
        "last_contact": days_from_today(0),
        "action": "Collect 10 vouchers",
    },
    {
        "sponsor": "Silog & Co.",
        "owner": "Vishnu Frendy",
        "channel": "Instagram",
        "stage": "Confirmed",
        "last_contact": days_from_today(0),
        "action": "Collect 2 vouchers",
    },
    {
        "sponsor": "Abad Volfe",
        "owner": "Anavi Verde",
        "channel": "Instagram",
        "stage": "Negotiating",
        "last_contact": days_from_today(-5),
    },
    {
        "sponsor": "Redbull",
        "owner": "Izy Ray",
        "channel": "Email",
        "stage": "Confirmed",
        "last_contact": days_from_today(-12),
        "action": "Collect 4 Redbull cases",
    },
]

# [CONSTANT] FOLLOW UP COLUMNS
FOLLOW_UP_COLUMNS = [
    ("sponsor", "Sponsor", 15),
    ("owner", "Owner", 15),
    ("channel", "Channel", 15),
    ("stage", "Stage", 15),
    ("last_contact", "Last Contact", 15),
    ("action", "Action Required", 30),
]


# [FUNCTION] GET FOLLOW UP ROWS
def get_follow_up_rows(today):
    rows = []

    for follow_up in FOLLOW_UPS:
        row = dict(follow_up)
        row["last_contact"] = format_date(follow_up["last_contact"])

        # Sponsors Still Being Chased Get A Calculated Action
        if follow_up["stage"] != CONFIRMED_STAGE:
            days = (today - follow_up["last_contact"]).days
            row["action"] = f"Follow up ({format_days(days)} no reply)"

        rows.append(row)

    return rows


# [FUNCTION] PRINT FOLLOW-UPS DUE TABLE ROW
def print_table_row(columns, row):
    line = ""

    for key, title, width in columns:
        line += f"{'|':2}{row[key]:{width}}"

    print(line + f"{'|':2}")


# [FUNCTION] PRINT FOLLOW-UPS DUE TABLE
def print_table(columns, rows, border_char="="):
    table_width = sum(width + 2 for key, title, width in columns) + 1
    border = border_char * table_width
    header = {key: title for key, title, width in columns}

    print(border)
    print_table_row(columns, header)
    print(border)

    for row in rows:
        print_table_row(columns, row)

    print(border)


# [CONSTANT] DEPARTMENT REQUESTS
DEPARTMENT_REQUESTS = [
    {
        "department": "Creatives",
        "title": "Tiangge Stall Decoration",
        "due": days_from_today(-2),
        "requested_by": "Vishnu Frendy",
        "note": "Needed for beautifying the stall",
        "status": "In progress",
    },
    {
        "department": "Creatives",
        "title": "Sponsor logos on Tiangge poster",
        "due": days_from_today(3),
        "requested_by": "Vishnu Frendy",
        "note": "Sedap Place and Silog & Co. logos",
        "status": "In progress",
    },
    {
        "department": "Events",
        "title": "IP65 leads for 2 vendor stalls",
        "due": days_from_today(5),
        "requested_by": "Anavi Verde",
        "note": "Required by Guild for powered stalls",
        "status": "Waiting",
    },
    {
        "department": "Marketing",
        "title": "Sedap Place shoutout story",
        "due": days_from_today(7),
        "requested_by": "Izy Ray",
        "note": "Sponsor deliverable for Tiangge",
        "status": "Not Started",
    },
]


# [FUNCTION] DESCRIBE DUE DATE
def describe_due_date(due_date, today):
    days_late = (today - due_date).days

    if days_late > 0:
        return f"Overdue by {format_days(days_late)}"

    if days_late == 0:
        return "Due today"

    return ""

# [FUNCTION] DESCRIBE REQUEST STATUS
def describe_request_status(request, today):
    if request["status"] == COMPLETE_STATUS:
        return COMPLETE_STATUS

    return describe_due_date(request["due"], today) or request["status"]


# [FUNCTION] GET REQUEST ROWS
def get_request_rows(today):
    rows = []

    for request in DEPARTMENT_REQUESTS:
        row = dict(request)
        row["due"] = format_date(request["due"])
        row["status"] = describe_request_status(request, today)
        rows.append(row)

    return rows


# [FUNCTION] GET WEEK RANGE
def get_week_range(today):
    # Calculate Monday (Start Of The Week) & Sunday (End Of The Week)
    start_of_week = today - timedelta(days=today.weekday())
    end_of_week = start_of_week + timedelta(days=6)
    return start_of_week, end_of_week


# [CONSTANT] CONFIRMED SPONSORS
CONFIRMED_SPONSORS = [
    {
        "sponsor": "Sedap Place",
        "event": "Tiangge Night Market",
        "item": "meal vouchers",
        "quantity": 10,
        "value": 15,
        "purpose": "raffle prizes",
        "confirmed": days_from_today(0),
        "channel": "Email",
        "next_step": "Collect vouchers from shop",
        "next_step_due": days_from_today(2),
    },
    {
        "sponsor": "Silog & Co.",
        "event": "Tiangge Night Market",
        "item": "meal vouchers",
        "quantity": 2,
        "value": 20,
        "purpose": "Best Stall prizes",
        "confirmed": days_from_today(0),
        "channel": "Instagram",
        "next_step": "Collect vouchers",
        "next_step_due": days_from_today(0),
    },
    {
        "sponsor": "Redbull",
        "event": "Tiangge Night Market",
        "item": "Redbull cases",
        "quantity": 4,
        "value": 0,
        "purpose": "volunteer drinks",
        "confirmed": days_from_today(-12),
        "channel": "Email",
        "next_step": "Collect Redbull cases",
        "next_step_due": days_from_today(1),
    },
]


# [FUNCTION] GET CONFIRMED THIS WEEK
def get_confirmed_this_week(start_of_week, end_of_week):
    return [item for item in CONFIRMED_SPONSORS if start_of_week <= item["confirmed"] <= end_of_week]


#[FUNCTION] COUNT CONTACTS LOGGED
def count_contacts_logged(start_of_week, end_of_week):
    # Contacts Logged In This Run Only Count If Their Date Falls In This Week
    logged_this_week = 0

    for contact in CONTACT_LOG:
        contact_date = datetime.strptime(contact["date"], DATE_FORMAT).date()

        if start_of_week <= contact_date <= end_of_week:
            logged_this_week += 1

    return sum(officer["contacts"] for officer in OFFICER_STATS) + logged_this_week


# [FUNCTION] DESCRIBE NEXT STEP
def describe_next_step(item, today):
    next_step = f"Next step: {item['next_step']} by {format_date(item['next_step_due'])}"
    due_message = describe_due_date(item["next_step_due"], today)

    return f"{next_step} ({due_message.lower()})" if due_message else next_step


# [FUNCTION] GET ALL OFFICER STATS
def get_all_officer_stats():
    # Every Listed Officer Is Included, Even With No Activity This Week
    saved_stats = {officer["officer"]: officer for officer in OFFICER_STATS}
    no_activity = {"contacts": 0, "follow_ups": 0, "confirmed": 0}

    return [saved_stats.get(name, {"officer": name, **no_activity}) for name in OFFICERS.values()]


# [FUNCTION] PRINT SUMMARY CARDS
def print_summary_cards(cards):
    border = "   ".join("-" * (width + 3) for title, value, width in cards)

    print(border)
    print("   ".join(f"| {title:{width}}|" for title, value, width in cards))
    print(border)
    print("   ".join(f"| {value:^{width}}|" for title, value, width in cards))
    print(border)


# [FUNCTION] CALCULATE OFFICER SCORE
def calculate_score(officer):
    return (
        officer["contacts"] * CONTACT_SCORE
        + officer["follow_ups"] * FOLLOW_UP_SCORE
        + officer["confirmed"] * CONFIRMED_SCORE
    )


# [FUNCTION] DISPLAY MAIN MENU
def display_main_menu(menu_options, width):
    border = "-" * width
    print(border)
    print(f"|{'Main Menu':^{width - 2}}|")
    print(border)

    for number, (label, action) in menu_options.items():
        print(f"| {f'{number}. {label}':{width - 3}}|")

    print(border)


# [FUNCTION] MAIN MENU
def main_menu(width=MAIN_MENU_WIDTH):
    menu_options = {
        "1": ("Log Contact", log_contact),
        "2": ("Add New Sponsor", add_new_sponsor),
        "3": ("Follow-ups Due", follow_ups_due),
        "4": ("Department Requests", department_requests),
        "5": ("Weekly PR Summary", weekly_pr_summary),
        "0": ("Exit", None),
    }

    # Display The Main Menu
    display_main_menu(menu_options, width)

    while True:
        # Input an option
        option = input("\nEnter option: ").strip()

        if option not in menu_options:
            print("\nInvalid choice. Please try again!")
            continue

        label, action = menu_options[option]

        if action is None:
            print("\nThank you for using CFSS Public Relations (PR) Hub!")
            break

        action()
        display_main_menu(menu_options, width)


# [FUNCTION] OPTION 1: LOG CONTACT
def log_contact(width=LOG_CONTACT_WIDTH):
    print_header()

    contact = collect_fields(LOG_CONTACT_FIELDS)
    CONTACT_LOG.append(contact)

    print("\n\nNew Log Contact Added!")
    border = "-" * width
    print(border)
    print(f"|{'New Log Contact':^{width - 2}}|")
    print(border)

    for key, label, prompt, rule in LOG_CONTACT_FIELDS:
        print(f"- {label}: {contact[key]}")

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print_header()
            return  # Goes back to the main menu loop that opened this option
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] OPTION 2: ADD NEW SPONSOR
def add_new_sponsor(width=NEW_SPONSOR_WIDTH):
    print_header()

    sponsor = collect_fields(NEW_SPONSOR_FIELDS)

    SPONSOR_LOG.append(sponsor)

    print("\n\nNew Sponsor Added!")
    border = "=" * width
    print(border)
    print(f"|{'New Sponsor':^{width - 2}}|")
    print(border)

    for key, label, prompt, rule in NEW_SPONSOR_FIELDS:
        print(f"- {label}: {sponsor[key]}")

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print_header()
            return # Goes back to the main menu loop that opened this option
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] OPTION 3: FOLLOW-UPS DUE
def follow_ups_due():
    print_header()
    print("\nFollow-Ups Due (All Officer)")
    print_table(FOLLOW_UP_COLUMNS, get_follow_up_rows(get_today()))

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * HEADER_WIDTH)
            return  # Goes back to the main menu loop that opened this option
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] OPTION 4: DEPARTMENT REQUESTS
def department_requests():
    print_header()
    print("\nDepartment Requests")
    print_table(DEPARTMENT_REQUEST_COLUMNS, get_request_rows(get_today()))

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * HEADER_WIDTH);
            print("\n")
            return # Goes back to the main menu loop that opened this option
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] OPTION 5: WEEKLY PR SUMMARY
def weekly_pr_summary():
    print_header()
    print("\nPR Weekly Summary")

    # Get Today's Date
    today = get_today()

    # Calculate Monday (Start Of The Week) & Sunday (End Of The Week)
    start_of_week, end_of_week = get_week_range(today)

    # Display Week Range
    print(f"Week Range: {start_of_week} to {end_of_week}\n")

    # Calculate The Summary Values
    confirmed_this_week = get_confirmed_this_week(start_of_week, end_of_week)
    contacts_logged = count_contacts_logged(start_of_week, end_of_week)
    sponsors_confirmed = len(confirmed_this_week)
    value_secured = sum(item["quantity"] * item["value"] for item in confirmed_this_week)
    follow_ups_count = sum(1 for follow_up in FOLLOW_UPS if follow_up["stage"] != CONFIRMED_STAGE)

    print_summary_cards([
        ("Contacts Logged", str(contacts_logged), 16),
        ("Sponsors Confirmed", str(sponsors_confirmed), 19),
        ("Value Secured", f"${value_secured}", 14),
        ("Follow-Ups Due", str(follow_ups_count), 15),
    ])

    print("\n\nConfirmed this week")

    if not confirmed_this_week:
        print("\nNo sponsors were confirmed this week.")

    for item in confirmed_this_week:
        total = item["quantity"] * item["value"]

        print_box([
            f"{item['sponsor']} - {item['event']}",
            f"{item['quantity']} x ${item['value']} {item['item']} (${total} total) for {item['purpose']}",
            f"Confirmed {format_date(item['confirmed'])} via {item['channel']}",
            describe_next_step(item, today),
        ])

    print("\n\nWaiting on other departments")

    departments = dict.fromkeys(request["department"] for request in DEPARTMENT_REQUESTS)

    for department in departments:
        print(f"\n[{department}]")

        for request in DEPARTMENT_REQUESTS:
            if request["department"] == department:
                print_box([
                    request["title"],
                    f"Due {format_date(request['due'])} . Requested by {request['requested_by']} . {request['note']}",
                ])

    print("\n\nOfficer contribution ranking")

    ranking = sorted(get_all_officer_stats(), key=calculate_score, reverse=True)
    ranking_rows = []
    rank = 0
    previous_score = None

    for position, officer in enumerate(ranking, start=1):
        score = calculate_score(officer)

        # Officers With Equal Scores Share The Same Rank
        if score != previous_score:
            rank = position
            previous_score = score

        ranking_rows.append({
            "rank": str(rank),
            "officer": officer["officer"],
            "contacts": str(officer["contacts"]),
            "follow_ups": str(officer["follow_ups"]),
            "confirmed": str(officer["confirmed"]),
            "score": str(score),
        })

    print_table(RANKING_COLUMNS, ranking_rows, border_char="-")
    print(
        f"Score = (Contacts × {CONTACT_SCORE}) + "
        f"(Follow-ups completed × {FOLLOW_UP_SCORE}) + "
        f"(Sponsors confirmed × {CONFIRMED_SCORE})"
    )

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * HEADER_WIDTH)
            print("\n")
            return  # Goes back to the main menu loop that opened this option
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] SELECT OFFICER
def select_officer():
  while True:
    # Ask "Who is currently using the system?"
    who = input("\nWho is currently using the system?: ").strip()

    try:
      return OFFICERS[who]
    except KeyError:
      print("\nInvalide choice. Please enter a number from the list.")


# [FUNCTION] MAIN
def main():
    pr_officer_list()

    # Ask who is currently using the system
    officer = select_officer()

    # Display header
    print_header()

    # Display "Welcome, officer_name!" & main menu
    print(f"\n\nWelcome, {officer}!")
    main_menu()

if __name__ == "__main__":
    main()
```
