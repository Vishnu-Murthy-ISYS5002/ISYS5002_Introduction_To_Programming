```python
from datetime import datetime, timedelta


# [CONSTANTS]
OFFICERS = {"1": "Vishnu Frendy", "2": "Anavi Verde", "3": "Erich Corin", "4": "Izy Ray", "5": "Elysha Miles", "6": "Owen Mendoza",}
HEADER_WIDTH = 62
OFFICER_LIST_WIDTH = 23
BOX_WIDTH = 85


# [FUNCTION] PAGE HEADER
def print_header():
    print(("=" * HEADER_WIDTH) + "\n")
    print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")


# [FUNCTION] PUBLIC RELATIONS (PR) OFFICER LIST
def pr_officer_list(width=OFFICER_LIST_WIDTH):
    print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")

    # Display The PR Officer List
    border = "-" * width
    print(border)
    print(f"|{'PR Officer List':^{width - 2}}|")
    print(border)

    for number, officer in OFFICERS.items():
        print(f"| {f'{number}. {officer}':{width - 3}}|")

    print(border)


# PRINT BOX LINES
def print_box(lines, width=BOX_WIDTH):
    border = "-" * width
    print(border)

    for line in lines:
        print(f"{'|':2}{line:{width - 3}}{'|':2}")

    print(border)


# [FUNCTION] MAIN MENU
def main_menu():
    # Display The Main Menu
    print("-" * 29)
    print("|         " + "Main Menu" + "         |")
    print("-" * 29)
    print("| " + "1. Log Contact" + "            |")
    print("| " + "2. Add New Sponsor" + "        |")
    print("| " + "3. Follow-ups Due" + "         |")
    print("| " + "4. Department Requests" + "    |")
    print("| " + "5. Weekly PR Summary" + "      |")
    print("| " + "0. Exit" + "                   |")
    print("-" * 29)

    while True:
        # Input an option
        option = input("\nEnter option: ")

        if option == "1":
            log_contact()

        elif option == "2":
            add_new_sponsor()

        elif option == "3":
            follow_ups_due()

        elif option == "4":
            department_requests()

        elif option == "5":
            weekly_pr_summary()

        elif option == "0":
            print("\nThank you for using CFSS Public Relations (PR) Hub!")
            break

        else:
            print("\nInvalid choice. Please try again!")

        if __name__ == "main_menu__":
            main_menu()


# OPTION 1: LOG CONTACT
def log_contact():
    print("=" * 62)
    print("\n"); print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
    sponsor = input("Enter sponsor name: ")
    communication_channel = input("Enter communication channel (Email/Instagram): ")
    date = input("Enter date (dd/mm/yyyy): ")
    contribution_type = input("Enter contribution type (Voucher/Merch): ")
    contribution_quantity = input("Enter quantity: ")
    event = input("Enter event name: ")

    print("\n\nNew Log Contact Added!")
    print("-" * 27)
    print("|     " + "New Log Contact" + "     |")
    print("-" * 27)
    print("- Sponsor Name: " + sponsor)
    print("- Communication Channel: " + communication_channel)
    print("- Date: " + date)
    print("- Contribution Type: " + contribution_type)
    print("- Quantity: " + contribution_quantity)
    print("- Event Name: " + event)

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * 62); print("\n")
            print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
            main_menu()
            break  # Exits the sub_menu loop and goes back to main()
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# OPTION 2: ADD NEW SPONSOR
def add_new_sponsor():
    print("=" * 62)
    print("\n"); print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
    sponsor_business_name = input("Enter sponsor business name: ")
    sponsor_business_type = input("Enter sponsor business type: ")
    sponsor_contact_person = input("Enter sponsor contact person: ")
    sponsor_job_title = input("Enter sponsor job title: ")
    sponsor_email_address = input("Enter sponsor email address: ")
    sponsor_phone_number = input("Enter sponsor phone number: ")
    sponsor_initial_contact_date = input("Enter sponsor initial contact date (dd/mm/yyyy): ")

    print("\n\nNew Sponsor Added!")
    print("=" * 23)
    print("|     " + "New Sponsor" + "     |")
    print("=" * 23)
    print("- Business name: " + sponsor_business_name)
    print("- Business type: " + sponsor_business_type)
    print("- Contact person: " + sponsor_contact_person)
    print("- Job title: " + sponsor_job_title)
    print("- Email address: " + sponsor_email_address)
    print("- Phone number: " + sponsor_phone_number)
    print("- Initial contact date: " + sponsor_initial_contact_date)

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * 62); print("\n")
            print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
            main_menu()
            break  # Exits the sub_menu loop and goes back to main()
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# OPTION 3: FOLLOW-UPS DUE
def follow_ups_due():
    print("=" * 62)
    print("\n"); print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
    print("\nFollow-Ups Due (All Officer)")
    print("=" * 118)
    print(f"{"|":2}{"Sponsor":15}{"|":2}{"Owner":15}{"|":2}{"Channel":15}{"|":2}{"Stage":15}{"|":2}{"Last Contact":15}{"|":2}{"Action Required":30}{"|":2}")
    print("=" * 118)
    print(f"{"|":2}{"Kenidonuts":15}{"|":2}{"Vishnu Frendy":15}{"|":2}{"Email":15}{"|":2}{"Contacted":15}{"|":2}{"08 Oct 2026":15}{"|":2}{"Follow up (7 days no reply)":30}{"|":2}")
    print(f"{"|":2}{"Silog & Co.":15}{"|":2}{"Vishnu Frendy":15}{"|":2}{"Instagram":15}{"|":2}{"Confirmed":15}{"|":2}{"04 Oct 2026":15}{"|":2}{"Collected 10 vouchers":30}{"|":2}")
    print(f"{"|":2}{"Abad Volfe":15}{"|":2}{"Anavi Verde":15}{"|":2}{"Instagram":15}{"|":2}{"Negotiating":15}{"|":2}{"10 Oct 2026":15}{"|":2}{"Follow up (5 days no reply)":30}{"|":2}")
    print(f"{"|":2}{"Redbull":15}{"|":2}{"Izy Ray":15}{"|":2}{"Email":15}{"|":2}{"Confirmed":15}{"|":2}{"12 Oct 2026":15}{"|":2}{"Collected 4 Redbull cases":30}{"|":2}")
    print("=" * 118)

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * 62)
            main_menu()
            break  # Exits the sub_menu loop and goes back to main()
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# OPTION 4: DEPARTMENT REQUESTS
def department_requests():
    print("=" * 62)
    print("\n"); print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
    print("\nDepartment Requests")
    print("=" * 94)
    print(f"{"|":2}{"Department":15}{"|":2}{"Requests":35}{"|":2}{"Due":15}{"|":2}{"Status":20}{"|":2}")
    print("=" * 94)
    print(f"{"|":2}{"Creatives":15}{"|":2}{"Tiangge stall decoration":35}{"|":2}{"14 Oct 2026":15}{"|":2}{"Overdue by 2 days":20}{"|":2}")
    print(f"{"|":2}{"Creatives":15}{"|":2}{"Sponsor logos on Tiangge poster":35}{"|":2}{"11 Oct 2026":15}{"|":2}{"In progress":20}{"|":2}")
    print(f"{"|":2}{"Events":15}{"|":2}{"IP65 leads for 2 vendor stalls":35}{"|":2}{"09 Oct 2026":15}{"|":2}{"Waiting":20}{"|":2}")
    print(f"{"|":2}{"Marketing":15}{"|":2}{"Sedap Place shoutout story":35}{"|":2}{"07 Oct 2026":15}{"|":2}{"Not Started":20}{"|":2}")
    print("=" * 94)

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * 62); print("\n")
            main_menu()
            break  # Exits the sub_menu loop and goes back to main()
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# OPTION 5: WEEKLY PR SUMMARY
def weekly_pr_summary():
    print("=" * 62)
    print("\n"); print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
    print("\nPR Weekly Summary")

    # Get Today's Date
    today = datetime.today().date()

    # Calculate Monday (Start Of The Week) & Sunday (End Of The Week)
    start_of_week = today - timedelta(days=today.weekday())
    end_of_week = start_of_week + timedelta(days=6)

    # Display Week Range
    print(f"Week Range: {start_of_week} to {end_of_week}\n")

    print("-" * 19, "-" * 22, "-" * 17, "-" * 18, sep="   ")
    print(f"{"|":2}{"Contacts Logged":16}{"|":4}{"|":2}{"Sponsors Confirmed":19}{"|":4}{"|":2}{"Value Secured":14}{"|":4}{"|":2}{"Follow-Ups Due":15}{"|":2}")
    print("-" * 19, "-" * 22, "-" * 17, "-" * 18, sep="   ")
    print(f"{"|":2}{"11":^16}{"|":4}{"|":2}{"2":^19}{"|":4}{"|":2}{"$190":^14}{"|":4}{"|":2}{"3":^15}{"|":2}")
    print("-" * 19, "-" * 22, "-" * 17, "-" * 18, sep="   ")

    print("\n\nConfirmed this week")
    print_box([
      "Sedap Place - Tiangge Night Market",
      "10 x $15 meal vouchers ($150 total) for raffle prizes",
      "Confirmed 15 Oct 2026 via email",
      "Next step: Collect vouchers from shop by 18 Oct 2026",
    ])

    print_box([
      "Silog & Co. - Tiangge Night Market",
      "2 x $20 meal vouchers ($40 total) for Best Stall prizes",
      "Confirmed 14 Oct 2026 via Instagram",
      "Next step: Collect vouchers by 16 Oct 2026 (due today)",
    ])

    print("\n\nWaiting on other departments")
    print("[Creatives]")
    print_box([
      "Tiangge Stall Decoration",
      "Due 14 Oct 2026 . Requested by Vishnu Frendy . Needed for beautifying the stall",
    ])
    print_box([
      "Sponsor logos on Tiangge poster",
      "Due 11 Oct 2026 . Requested by Vishnu Frendy . Sedap Place and Silog & Co. logos",
    ])

    print("\n[Events]")
    print_box([
      "IP65 leads for 2 vendor stalls",
      "Due 09 Oct 2026 . Requested by Anavi Verde . Required by Guild for powered stalls",
    ])

    print("\n[Marketing]")
    print_box([
      "Sedap Place shoutout story",
      "Due 07 Oct 2026 . Requested by Izy Ray . Sponsor deliverable for Tiangge",
    ])

    print("\n\nOfficer contribution ranking")
    print("-" * 103)
    print(f"{"|":2}{"Rank":15}{"|":2}{"Officer":15}{"|":2}{"Contact":15}{"|":2}{"Follow-Ups":15}{"|":2}{"Confirmed":15}{"|":2}{"Score":15}{"|":2}")
    print("-" * 103)
    print(f"{"|":2}{"1":15}{"|":2}{"Vishnu Frendy":15}{"|":2}{"5":15}{"|":2}{"2":15}{"|":2}{"2":15}{"|":2}{"19":15}{"|":2}")
    print(f"{"|":2}{"2":15}{"|":2}{"Anavi Verde":15}{"|":2}{"4":15}{"|":2}{"3":15}{"|":2}{"0":15}{"|":2}{"10":15}{"|":2}")
    print(f"{"|":2}{"3":15}{"|":2}{"Izy Ray":15}{"|":2}{"2":15}{"|":2}{"1":15}{"|":2}{"0":15}{"|":2}{"4":15}{"|":2}")
    print("-" * 103)
    print("Score = (Contacts × 1) + (Follow-ups completed × 2) + (Sponsors confirmed × 5)")

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * 62); print("\n")
            main_menu()
            break  # Exits the sub_menu loop and goes back to main()
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] SELECT OFFICER
def select_officer():
  while True:
    # Ask "Who is currently using the system?"
    who = input("\nWho is currently using the system?: ").strip()

    try:
      return OFFICERS[who]
    except KeyError:
      print("\nInvalide choice. Please enter a number from the list.")


# [FUNCTION] MAIN
def main():
    pr_officer_list()

    # Ask who is currently using the system
    officer = select_officer()

    # Display header
    print_header()

    # Display "Welcome, officer_name!" & main menu
    print(f"\n\nWelcome, {officer}!")
    main_menu()

if __name__ == "__main__":
    main()
```
