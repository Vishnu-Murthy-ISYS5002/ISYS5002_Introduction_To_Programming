```python
import re
from datetime import datetime, timedelta


# [CONSTANTS]
HEADER_WIDTH = 62
OFFICER_LIST_WIDTH = 23
BOX_WIDTH = 85
LOG_CONTACT_WIDTH = 27
MAX_TEXT_LENGTH = 100
DATE_FORMAT = "%d/%m/%Y"
MIN_YEAR = 2000
MAX_QUANTITY = 100000
CHANNELS = ["Email", "Instagram"]
CONTRIBUTION_TYPES = ["Voucher", "Merch"]
NOT_PROVIDED = "Not provided"
MIN_PHONE_DIGITS = 8
MAX_PHONE_DIGITS = 15
CONTACT_LOG = []
MAIN_MENU_WIDTH = 29
NEW_SPONSOR_WIDTH = 23
SPONSOR_LOG = []
OFFICERS = {"1": "Vishnu Frendy", "2": "Anavi Verde", "3": "Erich Corin", "4": "Izy Ray", "5": "Elysha Miles", "6": "Owen Mendoza",}
LOG_CONTACT_FIELDS = {
    ("sponsor", "Sponsor Name", "Enter sponsor name: ", "text"),
    ("channel", "Communication Channel", "Enter communication channel (Email/Instagram): ", "channel"),
    ("date", "Date", "Enter date (dd/mm/yyyy): ", "date"),
    ("type", "Contribution Type", "Enter contribution type (Voucher/Merch): ", "contribution_type"),
    ("quantity", "Quantity", "Enter quantity: ", "quantity"),
    ("event", "Event Name", "Enter event name: ", "text"),
}
NEW_SPONSOR_FIELDS = [
    ("business_name", "Business name", "Enter sponsor business name: ", "text"),
    ("business_type", "Business type", "Enter sponsor business type: ", "text"),
    ("contact_person", "Contact person", "Enter sponsor contact person: ", "text"),
    ("job_title", "Job title", "Enter sponsor job title: ", "text"),
    ("email", "Email address", "Enter sponsor email address (Enter to skip): ", "email"),
    ("phone", "Phone number", "Enter sponsor phone number (Enter to skip): ", "phone"),
    ("initial_contact_date", "Initial contact date", "Enter sponsor initial contact date (dd/mm/yyyy): ", "date"),
]


# [FUNCTION] PAGE HEADER
def print_header():
    print(("=" * HEADER_WIDTH) + "\n")
    print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")


# [FUNCTION] PUBLIC RELATIONS (PR) OFFICER LIST
def pr_officer_list(width=OFFICER_LIST_WIDTH):
    print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")

    # Display The PR Officer List
    border = "-" * width
    print(border)
    print(f"|{'PR Officer List':^{width - 2}}|")
    print(border)

    for number, officer in OFFICERS.items():
        print(f"| {f'{number}. {officer}':{width - 3}}|")

    print(border)


# [FUNCTION] PRINT BOX LINES
def print_box(lines, width=BOX_WIDTH):
    border = "-" * width
    print(border)

    for line in lines:
        print(f"{'|':2}{line:{width - 3}}{'|':2}")

    print(border)


# [FUNCTION] GET VALID INPUT
def get_valid_input(prompt, rule):
    while True:
        try:
            return VALIDATORS[rule](input(prompt).strip())
        except ValueError as error:
            print(error)


# [FUNCTION] COLLECT FIELDS
def collect_fields(fields):
    record = {}

    for key, label, prompt, rule in fields:
        record[key] = get_valid_input(prompt, rule)

    return record


# [FUNCTION] TEXT INPUT VALIDATION
def validate_text(text):
    if not text:
        raise ValueError("This field cannot be blank.")

    if len(text) > MAX_TEXT_LENGTH:
        raise ValueError(f"Please use {MAX_TEXT_LENGTH} characters or fewer.")

    return text


# [FUNCTION] DATE INPUT VALIDATION
def validate_date(text):
    try:
        date = datetime.strptime(text, DATE_FORMAT).date()
    except ValueError:
        raise ValueError("Invalid date. Please use dd/mm/yyyy.") from None

    if date.year < MIN_YEAR:
        raise ValueError(f"Please enter a date from {MIN_YEAR} onwards.")

    if date > get_today():
        raise ValueError("This date cannot be in the future.")

    return date.strftime(DATE_FORMAT)


# [FUNCTION] QUANTITY INPUT VALIDATION
def validate_quantity(text):
    try:
        quantity = int(text)
    except ValueError:
        raise ValueError("Please enter a whole number.") from None

    if not 1 <= quantity <= MAX_QUANTITY:
        raise ValueError(f"Quantity must be between 1 and {MAX_QUANTITY}.")

    return quantity


# [FUNCTION] OPTION INPUT VALIDATION
def validate_option(text, options):
    for option in options:
        if text.lower() == option.lower():
            return option

    raise ValueError(f"Please enter one of these: {', '.join(options)}.")


# [FUNCTION] CHANNEL INPUT VALIDATION
def validate_channel(text):
    return validate_option(text, CHANNELS)


# [FUNCTION] CONTRIBUTION TYPE INPUT VALIDATION
def validate_contribution_type(text):
    return validate_option(text, CONTRIBUTION_TYPES)


# [FUNCTION] EMAIL INPUT VALIDATION
def validate_email(text):
    if not text:
        return NOT_PROVIDED

    if not re.fullmatch(r"[^@\s]+@[^@\s]+\.[^@\s]+", text):
        raise ValueError("Invalid email. Please use a format like name@example.com.")

    return text


# [FUNCTION] PHONE INPUT VALIDATION
def validate_phone(text):
    if not text:
        return NOT_PROVIDED

    digits = re.sub(r"[\s()+-]", "", text)

    if not digits.isdigit() or not MIN_PHONE_DIGITS <= len(digits) <= MAX_PHONE_DIGITS:
        raise ValueError(f"Phone numbers need {MIN_PHONE_DIGITS} to {MAX_PHONE_DIGITS} digits.")

    return text

# [CONSTANT] VALIDATORS
VALIDATORS = {
    "text": validate_text,
    "date": validate_date,
    "quantity": validate_quantity,
    "channel": validate_channel,
    "contribution_type": validate_contribution_type,
    "email": validate_email,
    "phone": validate_phone,
}


# [FUNCTION] GET TODAY'S LOCAL DATE
def get_today():
    return datetime.today().date()


# [FUNCTION] DISPLAY MAIN MENU
def display_main_menu(menu_options, width):
    border = "-" * width
    print(border)
    print(f"|{'Main Menu':^{width - 2}}|")
    print(border)

    for number, (label, action) in menu_options.items():
        print(f"| {f'{number}. {label}':{width - 3}}|")

    print(border)


# [FUNCTION] MAIN MENU
def main_menu(width=MAIN_MENU_WIDTH):
    menu_options = {
        "1": ("Log Contact", log_contact),
        "2": ("Add New Sponsor", add_new_sponsor),
        "3": ("Follow-ups Due", follow_ups_due),
        "4": ("Department Requests", department_requests),
        "5": ("Weekly PR Summary", weekly_pr_summary),
        "0": ("Exit", None),
    }

    # Display The Main Menu
    display_main_menu(menu_options, width)

    while True:
        # Input an option
        option = input("\nEnter option: ").strip()

        if option not in menu_options:
            print("\nInvalid choice. Please try again!")
            continue

        label, action = menu_options[option]

        if action is None:
            print("\nThank you for using CFSS Public Relations (PR) Hub!")
            break

        action()
        display_main_menu(menu_options, width)


# [FUNCTION] OPTION 1: LOG CONTACT
def log_contact(width=LOG_CONTACT_WIDTH):
    print_header()

    contact = collect_fields(LOG_CONTACT_FIELDS)
    CONTACT_LOG.append(contact)

    print("\n\nNew Log Contact Added!")
    border = "-" * width
    print(border)
    print(f"|{'New Log Contact':^{width - 2}}|")
    print(border)

    for key, label, prompt, rule in LOG_CONTACT_FIELDS:
        print(f"- {label}: {contact[key]}")

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print_header()
            return  # Goes back to the main menu loop that opened this option
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] OPTION 2: ADD NEW SPONSOR
def add_new_sponsor(width=NEW_SPONSOR_WIDTH):
    print_header()

    sponsor = collect_fields(NEW_SPONSOR_FIELDS)

    SPONSOR_LOG.append(sponsor)

    print("\n\nNew Sponsor Added!")
    border = "=" * width
    print(border)
    print(f"|{'New Sponsor':^{width - 2}}|")
    print(border)

    for key, label, prompt, rule in NEW_SPONSOR_FIELDS:
        print(f"- {label}: {sponsor[key]}")

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print_header()
            return # Goes back to the main menu loop that opened this option
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] OPTION 3: FOLLOW-UPS DUE
def follow_ups_due():
    print("=" * 62)
    print("\n"); print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
    print("\nFollow-Ups Due (All Officer)")
    print("=" * 118)
    print(f"{"|":2}{"Sponsor":15}{"|":2}{"Owner":15}{"|":2}{"Channel":15}{"|":2}{"Stage":15}{"|":2}{"Last Contact":15}{"|":2}{"Action Required":30}{"|":2}")
    print("=" * 118)
    print(f"{"|":2}{"Kenidonuts":15}{"|":2}{"Vishnu Frendy":15}{"|":2}{"Email":15}{"|":2}{"Contacted":15}{"|":2}{"08 Oct 2026":15}{"|":2}{"Follow up (7 days no reply)":30}{"|":2}")
    print(f"{"|":2}{"Silog & Co.":15}{"|":2}{"Vishnu Frendy":15}{"|":2}{"Instagram":15}{"|":2}{"Confirmed":15}{"|":2}{"04 Oct 2026":15}{"|":2}{"Collected 10 vouchers":30}{"|":2}")
    print(f"{"|":2}{"Abad Volfe":15}{"|":2}{"Anavi Verde":15}{"|":2}{"Instagram":15}{"|":2}{"Negotiating":15}{"|":2}{"10 Oct 2026":15}{"|":2}{"Follow up (5 days no reply)":30}{"|":2}")
    print(f"{"|":2}{"Redbull":15}{"|":2}{"Izy Ray":15}{"|":2}{"Email":15}{"|":2}{"Confirmed":15}{"|":2}{"12 Oct 2026":15}{"|":2}{"Collected 4 Redbull cases":30}{"|":2}")
    print("=" * 118)

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * 62)
            main_menu()
            break  # Exits the sub_menu loop and goes back to main()
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] OPTION 4: DEPARTMENT REQUESTS
def department_requests():
    print("=" * 62)
    print("\n"); print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
    print("\nDepartment Requests")
    print("=" * 94)
    print(f"{"|":2}{"Department":15}{"|":2}{"Requests":35}{"|":2}{"Due":15}{"|":2}{"Status":20}{"|":2}")
    print("=" * 94)
    print(f"{"|":2}{"Creatives":15}{"|":2}{"Tiangge stall decoration":35}{"|":2}{"14 Oct 2026":15}{"|":2}{"Overdue by 2 days":20}{"|":2}")
    print(f"{"|":2}{"Creatives":15}{"|":2}{"Sponsor logos on Tiangge poster":35}{"|":2}{"11 Oct 2026":15}{"|":2}{"In progress":20}{"|":2}")
    print(f"{"|":2}{"Events":15}{"|":2}{"IP65 leads for 2 vendor stalls":35}{"|":2}{"09 Oct 2026":15}{"|":2}{"Waiting":20}{"|":2}")
    print(f"{"|":2}{"Marketing":15}{"|":2}{"Sedap Place shoutout story":35}{"|":2}{"07 Oct 2026":15}{"|":2}{"Not Started":20}{"|":2}")
    print("=" * 94)

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * 62); print("\n")
            main_menu()
            break  # Exits the sub_menu loop and goes back to main()
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] OPTION 5: WEEKLY PR SUMMARY
def weekly_pr_summary():
    print("=" * 62)
    print("\n"); print("=" * 15, "=" * 15, sep=" CFSS Public Relations (PR) Hub ")
    print("\nPR Weekly Summary")

    # Get Today's Date
    today = datetime.today().date()

    # Calculate Monday (Start Of The Week) & Sunday (End Of The Week)
    start_of_week = today - timedelta(days=today.weekday())
    end_of_week = start_of_week + timedelta(days=6)

    # Display Week Range
    print(f"Week Range: {start_of_week} to {end_of_week}\n")

    print("-" * 19, "-" * 22, "-" * 17, "-" * 18, sep="   ")
    print(f"{"|":2}{"Contacts Logged":16}{"|":4}{"|":2}{"Sponsors Confirmed":19}{"|":4}{"|":2}{"Value Secured":14}{"|":4}{"|":2}{"Follow-Ups Due":15}{"|":2}")
    print("-" * 19, "-" * 22, "-" * 17, "-" * 18, sep="   ")
    print(f"{"|":2}{"11":^16}{"|":4}{"|":2}{"2":^19}{"|":4}{"|":2}{"$190":^14}{"|":4}{"|":2}{"3":^15}{"|":2}")
    print("-" * 19, "-" * 22, "-" * 17, "-" * 18, sep="   ")

    print("\n\nConfirmed this week")
    print_box([
      "Sedap Place - Tiangge Night Market",
      "10 x $15 meal vouchers ($150 total) for raffle prizes",
      "Confirmed 15 Oct 2026 via email",
      "Next step: Collect vouchers from shop by 18 Oct 2026",
    ])

    print_box([
      "Silog & Co. - Tiangge Night Market",
      "2 x $20 meal vouchers ($40 total) for Best Stall prizes",
      "Confirmed 14 Oct 2026 via Instagram",
      "Next step: Collect vouchers by 16 Oct 2026 (due today)",
    ])

    print("\n\nWaiting on other departments")
    print("[Creatives]")
    print_box([
      "Tiangge Stall Decoration",
      "Due 14 Oct 2026 . Requested by Vishnu Frendy . Needed for beautifying the stall",
    ])
    print_box([
      "Sponsor logos on Tiangge poster",
      "Due 11 Oct 2026 . Requested by Vishnu Frendy . Sedap Place and Silog & Co. logos",
    ])

    print("\n[Events]")
    print_box([
      "IP65 leads for 2 vendor stalls",
      "Due 09 Oct 2026 . Requested by Anavi Verde . Required by Guild for powered stalls",
    ])

    print("\n[Marketing]")
    print_box([
      "Sedap Place shoutout story",
      "Due 07 Oct 2026 . Requested by Izy Ray . Sponsor deliverable for Tiangge",
    ])

    print("\n\nOfficer contribution ranking")
    print("-" * 103)
    print(f"{"|":2}{"Rank":15}{"|":2}{"Officer":15}{"|":2}{"Contact":15}{"|":2}{"Follow-Ups":15}{"|":2}{"Confirmed":15}{"|":2}{"Score":15}{"|":2}")
    print("-" * 103)
    print(f"{"|":2}{"1":15}{"|":2}{"Vishnu Frendy":15}{"|":2}{"5":15}{"|":2}{"2":15}{"|":2}{"2":15}{"|":2}{"19":15}{"|":2}")
    print(f"{"|":2}{"2":15}{"|":2}{"Anavi Verde":15}{"|":2}{"4":15}{"|":2}{"3":15}{"|":2}{"0":15}{"|":2}{"10":15}{"|":2}")
    print(f"{"|":2}{"3":15}{"|":2}{"Izy Ray":15}{"|":2}{"2":15}{"|":2}{"1":15}{"|":2}{"0":15}{"|":2}{"4":15}{"|":2}")
    print("-" * 103)
    print("Score = (Contacts × 1) + (Follow-ups completed × 2) + (Sponsors confirmed × 5)")

    while True:
        choice = input("\nDo you want to return to the main menu? (y/n): ").strip().lower()
        if choice == 'y':
            print("\nReturning to main menu...")
            print("=" * 62); print("\n")
            main_menu()
            break  # Exits the sub_menu loop and goes back to main()
        elif choice == 'n':
            print("\nStaying in sub-menu.")
        else:
            print("\nInvalid input. Please enter 'y' or 'n'.")


# [FUNCTION] SELECT OFFICER
def select_officer():
  while True:
    # Ask "Who is currently using the system?"
    who = input("\nWho is currently using the system?: ").strip()

    try:
      return OFFICERS[who]
    except KeyError:
      print("\nInvalide choice. Please enter a number from the list.")


# [FUNCTION] MAIN
def main():
    pr_officer_list()

    # Ask who is currently using the system
    officer = select_officer()

    # Display header
    print_header()

    # Display "Welcome, officer_name!" & main menu
    print(f"\n\nWelcome, {officer}!")
    main_menu()

if __name__ == "__main__":
    main()
```
