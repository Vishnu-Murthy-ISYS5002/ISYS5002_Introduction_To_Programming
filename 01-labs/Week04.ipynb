```python
# Display the Menu & Get the User's Choice

def get_operation():
    while True:
        print("Simple Calculator Menu")
        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Modulus")
        choice = input("Enter your choice (1-5): ")

        if choice in ("1", "2", "3", "4", "5"):
            return choice
        else:
            print("Invalid choice. \n\nPlease enter a number from 1 to 5.")


# Get the Numbers

def get_number(prompt):
    while True:
        value = input(prompt)
        try:
            return float(value)
        except ValueError:
            print("Invalid input. Please enter a valid number.")


# Perform the Calculation

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Cannot divide by zero."
    else:
        return a / b

def modulus(a, b):
    if b == 0:
        return "Error: Cannot divide by zero."
    else:
        return a % b

def perform_calculation(operation, num1, num2):
    if operation == "1":
        return add(num1, num2)
    elif operation == "2":
        return subtract(num1, num2)
    elif operation == "3":
        return multiply(num1, num2)
    elif operation == "4":
        return divide(num1, num2)
    elif operation == "5":
        return modulus(num1, num2)
    else:
        return "Invalid choice."


# Display the Result / Main programme

def main():
    operation = get_operation()
    num1 = get_number("Enter the first number: ")
    num2 = get_number("Enter the second number: ")
    result = perform_calculation(operation, num1, num2)

    if isinstance(result, float) and result.is_integer():
        result = int(result)

    print("Result:", result)

main()
```
